# Vetores como dados

## Ideia central


- Um vetor representa uma observação
- Cada componente representa uma variável (feature)

## Exemplos conceituais

- Pessoa -> (idade, altura)
- Produto -> (preço, avaliação, vendas)

## Criando um vetor com NumPy

```python
import numpy as np

pessoa = np.array([25, 1.75])

pessoa
```

```python
# Vários vetores (matriz) de dados

dados = np.array([
    [30, 1.77],
    [17, 1.83],
    [24, 1.65]
])

dados
```

  # Plotando vetores no ℝ²

```python
import matplotlib.pyplot as plt
```

```python
def plotVectors(vecs, cols, alpha=1):
    """
    Plot set of vectors.

    Parameters
    ----------
    vecs : array-like
        Coordinates of the vectors to plot. Each vectors is in an array. For
        instance: [[1, 3], [2, 2]] can be used to plot 2 vectors.
    cols : array-like
        Colors of the vectors. For instance: ['red', 'blue'] will display the
        first vector in red and the second in blue.
    alpha : float
        Opacity of vectors

    Returns:

    fig : instance of matplotlib.figure.Figure
        The figure of the vectors
    """
    plt.figure()
    plt.axvline(x=0, color='#A9A9A9', zorder=0)
    plt.axhline(y=0, color='#A9A9A9', zorder=0)

    for i in range(len(vecs)):
        x = np.concatenate([[0,0],vecs[i]])
        plt.quiver([x[0]],
                   [x[1]],
                   [x[2]],
                   [x[3]],
                   angles='xy', scale_units='xy', scale=1, color=cols[i],
                  alpha=alpha)
```

```python
u = np.array([2, 3])
v = np.array([4, -1])

plotVectors([u, v], ['#FF9A13', '#1190FF'])

plt.xlim(-5, 5)
plt.ylim(-5, 5)
```

# Calculando o ângulo entre vetores

```python
def calcular_angulo(v1, v2):
  prod = v1.dot(v2)
  norma_v1 = np.linalg.norm(v1)
  norma_v2 = np.linalg.norm(v2)
  r = prod / (norma_v1 * norma_v2)
  ang = np.arccos(r)
  return 180 / np.pi * ang
```

```python
u = np.array([1, 0])
v = np.array([0, 1])

plotVectors([u, v], ['red', 'blue'])
plt.xlim(-1, 5)
plt.ylim(-1, 5)
```

```python
calcular_angulo(u, v)
```
